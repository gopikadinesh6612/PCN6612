# Wired TCP Results Graphs

This notebook reads `tcp_results.csv` generated by the matching C++ program and plots its values.

**Note:** The included C++ file uses illustrative hardcoded sample values, not actual NS-3 measurements.

```python
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

csv_path = Path('tcp_results.csv')
if not csv_path.exists():
    raise FileNotFoundError(f"{csv_path} not found. Compile and run the matching C++ program first, in this folder.")
df = pd.read_csv(csv_path)
df
```

```python
metrics = {
    'Sent, received and lost packets': ['Sent', 'Received', 'Lost'],
    'Throughput (Mbps)': ['ThroughputMbps'],
    'Average delay (seconds)': ['DelaySeconds'],
}
for chart_title, columns in metrics.items():
    ax = df[columns].iloc[0].plot(kind='bar', title=chart_title, rot=0)
    ax.set_ylabel('Value')
    plt.tight_layout()
    plt.show()
```
