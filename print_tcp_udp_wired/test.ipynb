```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("tcp_results.csv")
display(df)

# 1. Throughput
plt.figure()
plt.bar(df["Protocol"], df["Throughput"])
plt.title("TCP Throughput")
plt.ylabel("Throughput (Mbps)")
plt.show()

# 2. Packet Loss
plt.figure()
plt.bar(df["Protocol"], df["Lost"])
plt.title("TCP Packet Loss")
plt.ylabel("Packets Lost")
plt.show()

# 3. Average Delay
plt.figure()
plt.bar(df["Protocol"], df["Delay"] * 1000)
plt.title("TCP Average Delay")
plt.ylabel("Delay (ms)")
plt.show()
```
