# Wireless UDP — Illustrative Graphs

These charts use example values for practice only. The companion C++ file prints configuration statements; it does not run NS-3 or measure these values. Replace the values below with real measured results if you later run an NS-3 simulation.

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    'Protocol': ['UDP'],
    'ThroughputMbps': [3.8],
    'LostPackets': [50],
    'AverageDelayMs': [15]
}
df = pd.DataFrame(data)
display(df)
```

```python
# 1. Throughput graph
plt.figure()
plt.bar(df['Protocol'], df['ThroughputMbps'])
plt.xlabel('Protocol')
plt.ylabel('Throughput (Mbps)')
plt.title('Wireless UDP Throughput (Illustrative)')
plt.tight_layout()
plt.show()

# 2. Packet loss graph
plt.figure()
plt.bar(df['Protocol'], df['LostPackets'])
plt.xlabel('Protocol')
plt.ylabel('Packets lost')
plt.title('Wireless UDP Packet Loss (Illustrative)')
plt.tight_layout()
plt.show()

# 3. Average delay graph
plt.figure()
plt.bar(df['Protocol'], df['AverageDelayMs'])
plt.xlabel('Protocol')
plt.ylabel('Average delay (ms)')
plt.title('Wireless UDP Average Delay (Illustrative)')
plt.tight_layout()
plt.show()
```
