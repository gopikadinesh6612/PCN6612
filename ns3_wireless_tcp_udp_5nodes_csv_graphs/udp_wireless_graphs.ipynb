# UDP Wireless 5-Node Network

Run the corresponding NS-3 program first. This notebook reads `udp_wireless_results.csv` and plots throughput, packet loss, and average delay.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('udp_wireless_results.csv')
display(df)
```

```python
if df.empty:
    print('No matching flow rows found. Check the NS-3 run and CSV file.')
else:
    # Throughput
    plt.figure()
    plt.bar(df['FlowID'].astype(str), df['ThroughputMbps'])
    plt.xlabel('Flow ID')
    plt.ylabel('Throughput (Mbps)')
    plt.title('UDP Wireless 5-Node Network - Throughput')
    plt.tight_layout()
    plt.show()

    # Packet loss
    plt.figure()
    plt.bar(df['FlowID'].astype(str), df['LostPackets'])
    plt.xlabel('Flow ID')
    plt.ylabel('Packets lost')
    plt.title('UDP Wireless 5-Node Network - Packet Loss')
    plt.tight_layout()
    plt.show()

    # Average delay in milliseconds
    plt.figure()
    plt.bar(df['FlowID'].astype(str), df['AverageDelaySeconds'] * 1000)
    plt.xlabel('Flow ID')
    plt.ylabel('Average delay (ms)')
    plt.title('UDP Wireless 5-Node Network - Average Delay')
    plt.tight_layout()
    plt.show()
```
