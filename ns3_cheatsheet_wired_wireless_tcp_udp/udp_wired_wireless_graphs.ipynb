# UDP Wired vs Wireless Results

Run both matching NS-3 programs first. Keep both generated CSV files in the same folder as this notebook, then choose Run All.


```python
import pandas as pd
import matplotlib.pyplot as plt

wired = pd.read_csv('udp_wired_results.csv')
wireless = pd.read_csv('udp_wireless_results.csv')
wired['Network'] = 'Wired'
wireless['Network'] = 'Wireless'
df = pd.concat([wired, wireless], ignore_index=True)
display(df)
```

```python
if df.empty:
    print('No flow rows found. Check the CSV files and NS-3 runs.')
else:
    # Throughput comparison
    plt.figure()
    plt.bar(df['Network'], df['ThroughputMbps'])
    plt.xlabel('Network type')
    plt.ylabel('Throughput (Mbps)')
    plt.title('UDP: Throughput - Wired vs Wireless')
    plt.tight_layout()
    plt.show()

    # Packet loss percentage comparison
    plt.figure()
    plt.bar(df['Network'], df['LossPercent'])
    plt.xlabel('Network type')
    plt.ylabel('Packet loss (%)')
    plt.title('UDP: Packet Loss - Wired vs Wireless')
    plt.tight_layout()
    plt.show()

    # Average delay comparison
    plt.figure()
    plt.bar(df['Network'], df['AverageDelayMs'])
    plt.xlabel('Network type')
    plt.ylabel('Average delay (ms)')
    plt.title('UDP: Average Delay - Wired vs Wireless')
    plt.tight_layout()
    plt.show()
```
