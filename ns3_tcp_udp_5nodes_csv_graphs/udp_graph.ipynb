# UDP 5-Node Wired Network Results

This notebook reads `udp_results.csv` generated by the NS-3 simulation and plots throughput, average delay, and packet loss.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('udp_results.csv')
display(df)
```

```python
if df.empty:
    print('No flow records found. Check that the NS-3 program ran successfully and created the CSV file.')
else:
    # Throughput graph
    plt.figure()
    plt.bar(df['flow_id'].astype(str), df['throughput_mbps'])
    plt.xlabel('Flow ID')
    plt.ylabel('Throughput (Mbps)')
    plt.title('UDP 5-Node Wired Network: Throughput')
    plt.tight_layout()
    plt.show()

    # Average delay graph (convert seconds to milliseconds)
    plt.figure()
    plt.bar(df['flow_id'].astype(str), df['avg_delay_seconds'] * 1000)
    plt.xlabel('Flow ID')
    plt.ylabel('Average delay (ms)')
    plt.title('UDP 5-Node Wired Network: Average Delay')
    plt.tight_layout()
    plt.show()

    # Packet loss graph
    plt.figure()
    plt.bar(df['flow_id'].astype(str), df['lost_packets'])
    plt.xlabel('Flow ID')
    plt.ylabel('Lost packets')
    plt.title('UDP 5-Node Wired Network: Packet Loss')
    plt.tight_layout()
    plt.show()
```
